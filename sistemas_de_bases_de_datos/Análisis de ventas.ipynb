```python
# Importar las librerías necesarias
import pandas as pd
import sqlite3
```

```python
# Conexión a la base de datos
dbname = 'TareaII'
conn = sqlite3.connect(dbname + '.db')
```

```python
# Crear cursor y listar tablas en la base de datos
cur = conn.cursor()
cur. execute("SELECT name FROM sqlite_master WHERE type='table';")
print(cur.fetchall())
```

```python
# Definir una función para ejecutar consultas y mostrar los resultados en formato DataFrame
def read_df(cur,command):
    result = cur.execute(command)
    col_names = list(map(lambda x: x[0], cur.description))
    df_query = pd.DataFrame(result, columns =col_names)
    print(df_query) 
```

```python
# Mostrar todos los registros de la tabla 'boleta'
read_df(cur,"""SELECT *
               FROM boleta;""")
```

```python
# Mostrar todos los registros de la tabla 'cantidad'
read_df(cur,"""SELECT *
               FROM cantidad;""")
```

```python
# Mostrar todos los registros de la tabla 'cliente'
read_df(cur,"""SELECT *
               FROM cliente;""")
```

```python
# Calcular el promedio de monto de boletas por cliente
read_df(cur,"""SELECT id_cliente, AVG(Monto) AS promedio_de_las_ventas_de_las_boletas
               FROM boleta
               GROUP BY id_cliente;""")
```

```python
# Contar el total de boletas por id de cliente
read_df(cur,"""SELECT id_cliente, COUNT(id_boleta) AS total_de_boletas
               FROM boleta
               GROUP BY id_cliente;""")
```

```python
# Contar el número de boletas por nombre de cliente 
read_df(cur,"""SELECT cliente.cliente, COUNT(id_boleta) AS número_de_boletas
               FROM boleta
               INNER JOIN cliente ON cliente.id_cliente = boleta.id_cliente
               GROUP BY cliente.cliente;""")
```

```python
# Contar las boletas por categoría de cliente
read_df(cur,"""SELECT cliente.categoria, COUNT(id_boleta) AS cantidad_de_boletas_clientes
               FROM boleta
               INNER JOIN cliente ON cliente.id_cliente = boleta.id_cliente
               GROUP BY cliente.categoria;""")
```

```python
# Contar boletas de clientes 'prime' con id_boleta >= 100
read_df(cur,"""SELECT cliente.categoria, COUNT(id_boleta) AS cantidad_de_boletas_clientes
               FROM boleta
               INNER JOIN cliente ON cliente.id_cliente = boleta.id_cliente
               WHERE boleta.id_boleta >=100 AND cliente.categoria = 'prime';""")
```

```python
# Total de productos comprados por cada cliente
read_df(cur,"""SELECT cliente.id_cliente, SUM(cantidad) AS cantidad_productos_comprados
               FROM boleta
               INNER JOIN cantidad ON cantidad.id_boleta = boleta.id_boleta
               INNER JOIN cliente ON cliente.id_cliente = boleta.id_cliente
               GROUP BY cliente.cliente;""")
```
